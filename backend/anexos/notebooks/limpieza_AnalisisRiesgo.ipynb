```python
import pandas as pd
df = pd.read_csv('/content/AnalisisRiesgo_20082017.csv', sep=';', encoding='utf-8-sig')
df
```

```python
nulos = df.isnull().sum()
nulos
```

```python
df = df[df['nombre contaminante'] == 'pm25'].reset_index(drop=True)

print(f'Filas después del filtro PM2.5: {len(df)}')
df
```

```python
df_limpio = df.copy()

cols_text = ['tipo de rezago', 'municipio', 'grupo diagnóstico', 'nombre contaminante', 'sexo', 'edad', 'rezago']
for col in cols_text:
    df_limpio[col] = df_limpio[col].str.strip()
cols_num = [
    'beta del contaminante',
    'exponencial del beta',
    'limite inferior del exponencial del beta',
    'limite superior del exponencial del beta',
    'porcentaje del incremento del riesgo'
]
for col in cols_num:
    df_limpio[col] = df_limpio[col].str.replace(',', '.', regex=False).astype(float)
duplicadas = df_limpio.duplicated().sum()
print(f'Filas duplicadas: {duplicadas}')
df_limpio = df_limpio.drop_duplicates()

df_limpio.dtypes
```

```python
for col in cols_num:
    min_val = df_limpio[col].min()
    max_val = df_limpio[col].max()
    df_limpio[col + '_normalizado'] = (df_limpio[col] - min_val) / (max_val - min_val)

df_limpio[['beta del contaminante', 'beta del contaminante_normalizado',
           'porcentaje del incremento del riesgo', 'porcentaje del incremento del riesgo_normalizado']].head(8)
```

```python
cols_categoricas = ['tipo de rezago', 'municipio', 'grupo diagnóstico', 'sexo', 'edad']
for col in cols_categoricas:
    codigos, categorias = pd.factorize(df_limpio[col])
    df_limpio[col + '_cod'] = codigos
    print(f'{col}: {dict(enumerate(categorias))}')
```
