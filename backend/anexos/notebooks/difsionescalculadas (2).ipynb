```python
import pandas as pd
df = pd.read_csv('/content/DefunsionesCalculadas_2008-2017.csv',sep=';',encoding='utf-8-sig')
df
```

```python
nulos = df.isnull().sum()
nulos
```

```python
df_limpio = df.copy()
cols_text =['Municipio', 'Sexo', 'Edad', 'Capitulo', 'Grupo']
df_limpio['Muertes'] = (df_limpio['Muertes'].str.replace(',','.',regex=False).astype(float))
duplicadas = df_limpio.duplicated().sum()
df_limpio.drop_duplicates()
```

```python
min_val = df_limpio['Muertes'].min()
max_val = df_limpio['Muertes'].max()
df_limpio['Muertes_normalizado'] = (df_limpio['Muertes']- min_val) /(max_val - min_val)
```

```python
cols_categoricas =['Municipio','Sexo','Edad','Capitulo','Grupo']
for col in cols_categoricas:
    codigos, categorias = pd.factorize(df_limpio[col])
    df_limpio[col + '_cod'] = codigos
    print(f'{col}: {dict(enumerate(categorias))}')
```
